# Data Analysis and Visualization

```python
# Happiness score by regional indicator for 2021 

plt.figure(figsize=(15,8)) 
sns.boxplot(x='Regional indicator',y='Ladder score',data=df2) 
plt.xticks(rotation=90) 
plt.title('Happiness score by regional indicator for 2021') 
plt.ylabel('Happiness score') 
plt.xlabel('Regional indicator') 
plt.show() 
```

```python
# Happiness score by GDP per capita for 2021 

plt.figure(figsize=(10,6)) 
sns.scatterplot(x='Logged GDP per capita',y='Ladder score',data=df2) 
plt.title('Happiness score by GDP per capita for 2021') 
plt.xlabel('Logged GDP per capita') 
plt.ylabel('Happines score') 
plt.show() 
```

```python
# Happiness score by social support for 2021 

plt.figure(figsize=(10,6)) 
plt.scatter(df2['Social support'],df2['Ladder score']) 
plt.title('Happiness score by social support for 2021') 
plt.xlabel('Social support') 
plt.ylabel('Happiness score') 
plt.show() 
```

```python
# Happiness score distribution for 2020 

plt.figure(figsize=(10,6)) 
plt.hist(df1['Life Ladder'],bins=30,color='green', edgecolor='black', alpha=0.7,label='2020') 
plt.title('Happiness score distribution for 2020') 
plt.xlabel('Happiness score') 
plt.ylabel('Frequency') 
plt.legend() 
plt.show() 
```

```python
# Happiness score by Log GDP per capita for 2020 

plt.figure(figsize=(10,6)) 
plt.scatter(df1['Log GDP per capita'],df1['Life Ladder'],color='green') 
plt.title('Happiness score by Log GDP per capita for 2020') 
plt.xlabel('Log GDP per capita') 
plt.ylabel('Happiness score') 
plt.show() 
```

```python
# Happiness score by healthy life expectancy for 2020 

plt.figure(figsize=(10,6)) 
plt.scatter(df1['Healthy life expectancy at birth'],df1['Life Ladder'],color='green') 
plt.title('Happiness score by healthy life expectancy for 2020') 
plt.xlabel('Healthy life expectancy') 
plt.ylabel('Happiness score') 
plt.show() 
```

```python
# Violin plot for Happines Score by Regional Indicator for 2021 

plt.figure(figsize=(15,8)) 
sns.violinplot(x='Regional indicator',y='Ladder score',data=df2) 
plt.xticks(rotation=90) 
plt.title('Distribution of Happiness score by regional indicator for 2021') 
plt.ylabel('Happiness score') 
plt.xlabel('Regional indicator') 
plt.show() 
```

```python
# Bar plot to compare Happiness score by country for 2020 and 2021 

countries_2021=df1['Country name'].unique() 
countries_2020=df1['Country name'].unique()
avg_scores_2021=[df2[df2['Country name']==country]['Ladder score'].mean() for country in countries_2021]
avg_scores_2020=[df1[df1['Country name']==country]['Life Ladder'].mean() for country in countries_2020]

plt.figure(figsize=(15,8))
bar_width=0.35
index=np.arange(len(countries_2021)) 

plt.bar(index, avg_scores_2020, bar_width, label='2020') 
plt.bar(index+bar_width, avg_scores_2021, bar_width, label='2021') 
plt.title('Comparison of Happiness scores by country for 2020 and 2021') 
plt.xlabel('Country') 
plt.ylabel('Average Happiness score') 
plt.xticks(index+bar_width/2, countries_2021, rotation=90) 
plt.tight_layout() 
plt.legend() 
plt.show() 
```
